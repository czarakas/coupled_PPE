```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ds=pd.read_csv('Figure4_Percent_Explained.csv')
```

```python
variable_to_explain=ds['Unnamed: 0']
```

```python
species = ('Adelie', 'Chinstrap', 'Gentoo')
counts = {
    'EF': ds.EF,
    'Albedo': ds.Albedo,
    'ra':ds.ra,
    'Interactions':ds.Interactions
}
width = 0.6  # the width of the bars: can also be len(x) sequence

plt.rcParams['hatch.linewidth'] = 3.5

fig, ax = plt.subplots(figsize=(13,7))
ax.xaxis.grid(zorder=0)
colors=['tab:blue','tab:orange','tab:green']

left = 0
i=0
for var, count in counts.items():
    if var=='Interactions':
        p = ax.barh(variable_to_explain, count, width, label=var,left=left,
                   color='white',hatch='//',edgecolor='tab:blue',zorder=3)#, bottom=bottom) 
        # Plot the edge
        ax.barh(variable_to_explain, count, width, label=var,left=left,
                   color='none',edgecolor='k',zorder=3)
    else:
        p = ax.barh(variable_to_explain, count, width, label=var,left=left,
                   color=colors[i],edgecolor='k',zorder=3)#, bottom=bottom)
    left += count
    i=i+1
    #ax.bar_label(p, label_type='center')

ax.legend()
plt.xticks(np.arange(0,110,10))
plt.xlim([0,100])     
plt.tight_layout()
plt.savefig('Figure4.pdf')
```
